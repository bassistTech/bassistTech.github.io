Bassists have been fascinated with the "speed" of amplifiers and speakers, for as long as I can remember. The first thing that got me interested in how speakers work was a comment that "tens are fast and fifteens are deep, because physics." I wondered, what physics?

Looking into the speed issue, I dug up the Wikipedia article on the harmoic oscillator:

https://en.wikipedia.org/wiki/Harmonic_oscillator

Quoting from the article: The equation of motion is:

$$F = -kx - c \dfrac {dx}{dt} - m \dfrac{d^2x}{dt^2}$$

which can be rewritten into the form:

$$\dfrac {d^2 x}{dt^2} + 2 \zeta \omega_0 \dfrac {dx}{dt} + \omega_-^2x = 0$$

where

$\omega_0 = \sqrt{k/m}$ is called the "undamped angular frequency of the oscillator",

$\zeta = \dfrac c {2\sqrt{mk}}$ is called the "damping ratio"....

The Q factor of a damped oscillator is defined as

$Q = 2 \pi \times \dfrac {\text {energy stored}}{\text {energy lost per cycle}}$

Q is related to the damping ratio by $Q = \dfrac 1 {2 \zeta}$ ....

The time an oscillator needs to adapt to changed external conditions is of the order $\tau = 1/(\zeta \omega_0)$. In physics, the adaptation is called relaxation, and $\tau$ is called the relaxation time.

End of quotation. I'm going to manipulate these equations just a bit...

$$\tau = \dfrac 1 {\zeta \omega_0}$$

but

$$\zeta = \dfrac 1 {2 Q}$$

Thus,

$$\tau = \dfrac {2Q}{\omega_0}$$

The equation of motion for a sealed speaker, in the absence of inductance, is exactly of the form of a damped harmonic oscillator. Its frequency response curve is dictated by two parameters:

$f_0 = 2 \pi \omega_0$, which is the resonant frequency.

$Q_{tc}$ is the total system Q factor, equal to $Q_{ts}\sqrt{1 + V_{as}/V_{box}}$

The speaker is governed by the same response time constant, converted to units of frequency:

$$\tau = \dfrac {Q_{tc}}{\pi f_0}$$

This is just for a sealed speaker, not for a ported system, and neglects inductance, but shows a basic relationship for the "speed" of a speaker:

* A speaker with higher $f_0$ is "faster."

* A speaker with lower $Q_{tc}$ is "faster."

The most interesting observation to be made is that this rule is utterly independent of cone size, cone mass, or the number of cones, so long as a system has acceptable parameters for use as a bass speaker. It seems quite likely to me that coil inductance does not affect this relationship, since its effect comes at much higher frequencies than the typical bass region. I haven't worked out how the port enters into this equation yet.

Another observation is that there's no way to change the speed of a speaker without also changing its bass response. If the change is audible, it's impossible to say if it's because of speed or bass response. But "tens are fast and fifteens are deep because physics."

What else affects the speed of bass amplification?

* Acoustics. A speaker is inefficient -- a few percent at best. This means it's weakly coupled to the acoustics of the room. Your speaker can start the air moving, but can't stop it from bouncing around. Small rooms, and rooms with a lot of reflective surfaces, will make it sound like your sound is "slower."

* Your amp has a response speed. A modern power amp reacts almost instantaneously to the input signal, but filter circuits in the preamp -- especially the bass EQ -- have a transient response function of their own, probably resembling the one given above. Many amps designed for electric bass use have a "voicing" with pronounced bass response, which will also affect transient response.

* Amp clipping and voice coil over-excursion. Both of these temporarily increase the effective Q factor while the waveform is in clipping. The first, by reducing the damping factor. The second by reducing the motor strength (BL) factor. However, you amp might not be clipping! The smaller Class-D heads that I've tested all have built-in limiters, to keep them from going into clipping.

Is it audible?

This is the eternal question. It would be hard to guess from testimonials that have not controlled for the above factors.
